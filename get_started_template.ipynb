### Getting started with a new Repository on Mac

```python
brew install python ## If not available
brew install git ## If not available
```

#### Configure one-time set up for Git

```python
git config --global user.name "Vinod"
git config --global user.email "vinod@example.com"
```

This dictates the Author: 'Vinod' <vinod@example.com> field that is shown in the git logs/ commit logs

#### Create a project folder for the new project

```python
mkdir ai-engineering
cd ai-engineering
```

#### Initialise a Git Repository within the project folder

```python
git init 

### Expected output: Initialised empty Git repository
```

#### Create a virtual environment

```python
python3 -m venv .venv 
```

#### Activate the virtual environment

```python
source .venv/bin/activate
```

```python
### Expected output: (.venv) 
```

#### To deactivate the current virtual environment

```python
deactivate
```

#### Install the necessary libraries for the current virtual environment

```python
pip install notebook ipykernel pandas matplotlib numpy
```

#### Creating the first notebook & accessing it in vscode

```python
touch test_notebook.ipynb
```

```python
code test_notebook.ipynb
### Set the kernel to Python 3 (ipykernel) in the notebook interface
```

#### Create a .gitignore

```python
touch .gitignore
code .gitignore

echo ".venv/" >> .gitignore
echo ".ipython_checkpoints" >> .gitignore
echo "test_notebook.ipynb" >> .gitignore
```

#### To check if file is being tracked (If file is tracked, it'll be printed)

```python
git ls-files test_notebook.ipynb
```

#### To stop a file that is currently being tracked but keeping it locally

```python
git rm --cached test_notebook.ipynb
git commit -m "Remove test_notebook.ipynb from tracking and update .gitignore"
git push
```

#### Creating Git Commit

```python
git status 
git add . (Or specify files to add)
git commit -m "Initial commit with .gitignore and notebook"


### In parallel, create an empty repository on GitHub
```
